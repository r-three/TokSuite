This notebook shows how to process lm-eval results and generates all tables in the paper

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
from pathlib import Path
from toksuite.logging.plot_utils import MODEL_TO_COLOR, setup_styles
from lm_eval.tasks.toksuite.utils import (
    LATEX_TABLE_CATEGORIES,
    get_table_str,
    get_accuracy_table_str,
    get_canonical_performance_table_str,
    LATEX_TABLE_CATEGORIES_CANONICAL,
)

setup_styles()
%matplotlib inline
```

```python
BASE_RESULTS_DIR = Path().resolve().parent.joinpath("results/toksuite")
assert BASE_RESULTS_DIR.exists(), (
    f"Base results directory {BASE_RESULTS_DIR} does not exist. Please run the evaluation scripts to generate results before running this notebook or point to the correct directory."
)

OUTPUT_TABLE_FORMAT = (
    "latex"  # or "markdown" for markdown tables or "dataframe" for pandas DataFrames
)
# OUTPUT_TABLE_FORMAT = "dataframe"
# OUTPUT_TABLE_FORMAT = "markdown"
```

## Main Table

```python
results_paths = [
    *list(BASE_RESULTS_DIR.joinpath("toksuite_on_toksuite").rglob("results*.json"))
]
tab = get_table_str(results_paths, output_format=OUTPUT_TABLE_FORMAT)

print(tab) if OUTPUT_TABLE_FORMAT != "dataframe" else tab
```
